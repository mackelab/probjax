```python
from brian2 import *

def simulate_neuron_network(conductance_values, synaptic_weight):
    # Define neuron model
    tau = 10 * ms
    eqs = '''
    dv/dt = (1-v)/tau : 1
    '''

    # Create neurons
    neurons = NeuronGroup(3, eqs, threshold='v>0.8', reset='v=0', method='exact')

    # Define synaptic connections
    synapses = Synapses(neurons, neurons, 'w : 1', on_pre='v_post += w')
    synapses.connect(i=0, j=1)  # Neuron 0 connects to Neuron 1
    synapses.connect(i=1, j=2)  # Neuron 1 connects to Neuron 2

    # Set synaptic weights
    synapses.w = synaptic_weight

    # Set conductance values
    neurons.tau = conductance_values

    # Create monitors to record neuron activity
    mon_v = StateMonitor(neurons, 'v', record=True)
    spike_mon = SpikeMonitor(neurons)

    # Run simulation
    run(100 * ms)

    # Plot results
    figure(figsize=(12, 4))
    subplot(121)
    plot(mon_v.t / ms, mon_v.v[0], label='Neuron 0')
    plot(mon_v.t / ms, mon_v.v[1], label='Neuron 1')
    plot(mon_v.t / ms, mon_v.v[2], label='Neuron 2')
    xlabel('Time (ms)')
    ylabel('Membrane Potential')
    legend()

    subplot(122)
    plot(spike_mon.t / ms, spike_mon.i, 'o')
    xlabel('Time (ms)')
    ylabel('Neuron Index')
    show()

# Example usage:
conductance_values = [15 * nS, 20 * nS, 10 * nS]  # Replace with your desired conductance values
synaptic_weight = 0.2  # Replace with your desired synaptic weight
simulate_neuron_network(conductance_values, synaptic_weight)
```
