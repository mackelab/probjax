```python
x = jnp.ones(1000)
y = jnp.ones((10,))
def f(x):
    def scan_fn(carry, xi):
        return carry + xi, carry + xi
    
    return jax.lax.scan(scan_fn, 0., x)
```

```python
jax.make_jaxpr(f)(x)
```

```python
jax.vmap(jax.vmap(f, in_axes=(0,None)), in_axes=(None, 0))(x, y).shape
```

```python
jax.make_jaxpr(jax.grad(f))(jnp.squeeze(x))
```
