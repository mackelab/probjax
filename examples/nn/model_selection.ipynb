```python
def generate_data(key,num_samples, num_components):
    key, subkey = jrandom.split(key)
    means = jax.random.normal(key, (num_components, 1))
    eps = jax.random.normal(subkey, (num_samples, 1))
    xs = means + eps[:, None]
    return means, xs
```

```python
means, xs = generate_data(jax.random.PRNGKey(0), 10, 3)
```
