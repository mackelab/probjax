```python
import jax
import jax.numpy as jnp
import jax.scipy.stats as stats

import matpl

def smc(target_logpdf, num_particles, num_iterations):
    def resample(key, weights, particles):
        indices = jax.random.categorical(key, jnp.log(weights), shape=(num_particles,))
        return particles[indices]

    def propagate(particles):
        return jax.vmap(target_logpdf)(particles)

    key = jax.random.PRNGKey(0)
    particles = jax.random.normal(key, shape=(num_particles,))  # Initialize particles from a standard normal distribution
    weights = jnp.ones(num_particles) / num_particles  # Initialize weights uniformly

    for _ in range(num_iterations):
        key, subkey = jax.random.split(key)
        particles = resample(subkey, weights, particles)
        particles = propagate(particles)
        weights = jnp.ones(num_particles) / num_particles

    return particles, weights

# Example usage
def target_logpdf(x):
    return -0.5 * (x - 2)**2  # Target distribution: N(0, 1)

num_particles = 1000
num_iterations = 2

particles, weights = smc(target_logpdf, num_particles, num_iterations)
```
