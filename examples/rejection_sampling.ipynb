```python
import numpy as np

def rejection_sampling(pdf_func, sample_func, num_samples, x_min, x_max, y_max):
    samples = []
    while len(samples) < num_samples:
        x = np.random.uniform(x_min, x_max)
        y = np.random.uniform(0, y_max)
        if y <= pdf_func(x):
            samples.append(sample_func(x))
    return np.array(samples)
```

```python
def gaussian_pdf(x):
    return np.exp(-0.5 * x**2) / np.sqrt(2 * np.pi)

samples = rejection_sampling(gaussian_pdf, np.random.normal, 1000, -5, 5, 0.4)
```
