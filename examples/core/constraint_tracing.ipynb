```python
import jax 
import jax.numpy as jnp

import matplotlib.pyplot as plt

v = jnp.array([[1, 2, 3], [4, 5, 0], [7, 3,1]])

# Mask
mask = jnp.array([[True, False, True], [False, True, False], [False, False, True]])

plt.imshow(mask)
```

```python
mask_indices1, mask_indices2 = jnp.mask_indices(3, lambda x,y: mask)
```

```python
collected_vs1 = v[mask_indices1]
collected_vs2 = v[mask_indices2]
```

```python
jnp.sum(collected_vs1 * collected_vs2, axis=-1)
```

```python
v@v.T
```

```python
plt.imshow(jax.nn.softmax(jnp.where(mask, v@v.T, -1e30), -1))
```

```python
plt.imshow(jax.nn.softmax(v@v.T, -1)*mask)
```
