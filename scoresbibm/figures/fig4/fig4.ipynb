```python
import svgutils.transform as sg
import sys 

#create new SVG figure
fig = sg.SVGFigure(width="10cm", height="11cm")

# load matpotlib-generated figures
fig1 = sg.fromfile('sir_example1.svg')
fig2 = sg.fromfile('sir_example_emulation.svg')
# fig12_legend = sg.fromfile('method_legend.svg')
# fig3 = sg.fromfile('conditional_two_moons.svg')

# get the plot objects
plot1 = fig1.getroot()
plot2 = fig2.getroot()

plot1.scale_xy(0.88, 0.88)
plot2.scale_xy(0.88, 0.88)
plot1.moveto(0, 15)
plot2.moveto(0, 220)



# add text labels
txt1 = sg.TextElement(5,15, "a", size=12, weight="bold", font="Arial")
txt2 = sg.TextElement(5,210, "b", size=12, weight="bold", font="Arial")


# append plots and labels to figure
fig.append([plot1, plot2])
fig.append([txt1, txt2])

fig.save("fig4.svg")
```
