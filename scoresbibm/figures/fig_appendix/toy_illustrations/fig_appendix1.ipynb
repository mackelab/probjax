```python
import svgutils.transform as sg
import sys 

#create new SVG figure
fig = sg.SVGFigure(width="20cm", height="9cm")

# load matpotlib-generated figures
fig1 = sg.fromfile('scoresbi_constraint_conditioning.svg')
fig2 = sg.fromfile('scoresbi_prior_change.svg')
fig3 = sg.fromfile('scoresbi_prior_change2.svg')
fig4 = sg.fromfile('scoresbi_likelihood_change.svg')
fig5 = sg.fromfile('scoresbi_likelihood_change2.svg')

# get the plot objects
plot1 = fig1.getroot()
plot2 = fig2.getroot()
plot3 = fig3.getroot()
plot4 = fig4.getroot()
plot5 = fig5.getroot()
plot1.moveto(20, 20)
plot2.moveto(340, 20)
plot3.moveto(340, 185)
plot4.moveto(560, 20)
plot5.moveto(560, 185)


# add text labels
txt1 = sg.TextElement(5,15, "a", size=12, weight="bold", font="Arial")
txt2 = sg.TextElement(320,15, "b", size=12, weight="bold", font="Arial")
txt3 = sg.TextElement(540,15, "c", size=12, weight="bold", font="Arial")

# append plots and labels to figure
fig.append([plot1, plot2,plot3,plot4,plot5])
fig.append([txt1,txt2,txt3])

fig.save("fig1_appendix.svg")
```
