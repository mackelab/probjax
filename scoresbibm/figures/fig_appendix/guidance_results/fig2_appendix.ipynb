```python
import svgutils.transform as sg
import sys 

#create new SVG figure
fig = sg.SVGFigure(width="19cm", height="8cm")

# load matpotlib-generated figures
fig1 = sg.fromfile('guidance_two_moons_example.svg')
fig2 = sg.fromfile('abstract_guidance_two_moons_example.svg')


# get the plot objects
plot1 = fig1.getroot()
plot2 = fig2.getroot()

plot1.moveto(20, 5)
plot2.moveto(20, 155)



# add text labels
txt1 = sg.TextElement(5,15, "a", size=12, weight="bold", font="Arial")
txt2 = sg.TextElement(5,175, "b", size=12, weight="bold", font="Arial")


# append plots and labels to figure
fig.append([plot1, plot2])
fig.append([txt1,txt2])

fig.save("fig2_appendix.svg")
```
