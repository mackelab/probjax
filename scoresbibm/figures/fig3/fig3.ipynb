```python
import svgutils.transform as sg
import sys 

#create new SVG figure
fig = sg.SVGFigure(width="10cm", height="9.5cm")

# load matpotlib-generated figures
fig1 = sg.fromfile('lv_few_observations.svg')
fig2 = sg.fromfile('lv_many_observations.svg')
fig3 = sg.fromfile('bm_lv.svg')
fig3b = sg.fromfile('bm_lv_posterior.svg')
fig3_legend = sg.fromfile("method_legend.svg")
# fig12_legend = sg.fromfile('method_legend.svg')
# fig3 = sg.fromfile('conditional_two_moons.svg')

# get the plot objects
plot1 = fig1.getroot()
plot2 = fig2.getroot()
plot3 = fig3.getroot()
plot3_legend = fig3_legend.getroot()
plot3b = fig3b.getroot()
# plot4 = fig3.getroot()
plot1.moveto(0, 5)
plot2.moveto(0, 120)
plot3.moveto(0, 250)
plot3_legend.moveto(260, 255)
plot3b.moveto(140, 250)
# plot3.moveto(-20, 190)
# plot4.moveto(540, 15)

# add text labels
txt1 = sg.TextElement(5,15, "a", size=12, weight="bold", font="Arial")
txt2 = sg.TextElement(5,130, "b", size=12, weight="bold", font="Arial")
txt3 = sg.TextElement(5,255, "c", size=12, weight="bold", font="Arial")

# append plots and labels to figure
fig.append([plot1, plot2, plot3, plot3b,plot3_legend])
fig.append([txt1, txt2,txt3])

fig.save("fig3.svg")
```
